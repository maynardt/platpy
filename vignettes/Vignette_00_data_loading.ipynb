<h2>
  Vignette 00: Loading Raw 10x Data into platpy<br>
</h2>

This vignette converts raw 10x Genomics output directories into platpy's native format.
Run it once per dataset; all subsequent vignettes load from the saved bundles in seconds.

**What this vignette does:**
- **Part 1 — Visium HD**: loads spot data at all bin levels (2/4/8/16 µm), attaches full-resolution IF images, and saves a platpy bundle.
- **Part 2 — Xenium**: loads the full-slide master, extracts each tissue section as its own named sample, and saves each bundle separately.

After running this vignette, update the `save_path` / `save_dir` variables in the subsequent vignettes to point at these saved bundles.

#### Note: As a first test of platpy, it's recommended that you skip this step and simply use the pre-made bundles that have been uploaded to Zenodo.  The following vignettes will automatically download the sample datasets for you.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
import os

from platpy.io_utils import load_sample, save_sample
from platpy.visium_loader import load_visium_directory, add_visium_fullres_channel
from platpy.xenium_loader import load_xenium
from platpy.channels import SampleSet
from platpy.plot_utils import plot_image_grid
```

---
### Part 1 — Visium HD

Point `visium_dir` at the SpaceRanger `outs/` folder.  
Optionally provide paths to full-resolution IF TIFFs — these are added as extra ImageChannels aligned to the same micron coordinate system.

```python
# --- UPDATE THESE PATHS ---
visium_dir        = "U:/Visium/visium_merge_1AD/merge_A/outs"
fullres_dapi_path = "U:/Visium/visium_merge_1AD/merge_A_fullres/merge_A_fullres_dapi.tif"
fullres_neun_path = "U:/Visium/visium_merge_1AD/merge_A_fullres/merge_A_fullres_Rbfox3.tif"
visium_save_path  = Path.home() / 'platpy_data' / 'VisiumSample'

if visium_save_path.exists():
    print("Visium bundle already exists — skipping. Delete the folder to re-process.")
    print(f"  {visium_save_path}")
else:
    print("Loading Visium from raw 10x files (this takes a few minutes)...")
    sample_vis = load_visium_directory(
        base_dir=visium_dir, name="VisiumSample", bin_sizes=[2, 4, 8, 16], debug=True
    )

    if os.path.exists(fullres_dapi_path):
        add_visium_fullres_channel(sample_vis, path=fullres_dapi_path, name="dapi")
    if os.path.exists(fullres_neun_path):
        add_visium_fullres_channel(sample_vis, path=fullres_neun_path, name="neun")

    save_sample(sample=sample_vis, out_prefix=visium_save_path, overwrite=False)
    print(f"\n✅ Saved to {visium_save_path}")
    print(f"   Omic channels : {list(sample_vis.omics.keys())}")
    print(f"   Image channels: {list(sample_vis.images.keys())}")
```

---
### Part 2 — Xenium

Point `xenium_dir` at the 10x Xenium output folder for the full slide.  
Define each tissue section as an axis-aligned bounding box in microns (`sample_rois`).  
The master slide is loaded once, then each section is extracted and saved as its own bundle.

**Tip:** Run the initial viewport inspection in Vignette 04 first (with `load_xenium` directly)
to identify the bounding boxes before committing them here.

```python
# --- UPDATE THESE PATHS ---
xenium_dir = "U:/Xenium/output-XETG00181__0071270__WT__20250908__180623"
save_dir   = Path.home() / 'platpy_data' / 'XeniumSamples'

# Bounding boxes for each tissue section: (min_x, max_x, min_y, max_y) in microns.
# Determined from initial inspection of the full slide in Vignette 04.
sample_rois = {
    'WT_01': (4120,  6975,   58, 4252),
    'WT_02': (7651, 10500,  491, 4481),
}

if all((save_dir / name).exists() for name in sample_rois):
    print("Xenium bundles already exist — skipping. Delete the folders to re-process.")
    for name in sample_rois:
        print(f"  {save_dir / name}")
else:
    print("Loading master Xenium slide (this takes ~10 minutes)...")
    master = load_xenium(xenium_dir)
    print(f"  Images: {list(master.images.keys())}")
    print(f"  Total transcripts: {master.omics['transcripts'].data.shape[0]:,}")

    # Save a lightweight lowres overview for slide-level navigation
    save_dir.mkdir(parents=True, exist_ok=True)
    overview = SampleSet(name='overview')
    overview.images['DAPI'] = master.images['DAPI']
    save_sample(overview, save_dir / 'overview', overwrite=True)

    # Extract and save each section
    for name, (min_x, max_x, min_y, max_y) in sample_rois.items():
        print(f"Extracting {name}...")
        s = master.extract_roi(min_x, max_x, min_y, max_y, new_name=name)
        save_sample(s, save_dir / name, overwrite=True)
        n_tx = s.omics['transcripts'].data.shape[0]
        print(f"  → {n_tx:,} transcripts  |  saved to {save_dir / name}")

    print("\n✅ All Xenium bundles saved.")
```
